# Technology Trends Data Analysis

## Coursera / IBM Data Analyst Capstone Project

This notebook analyzes a sample of Stack Overflow Developer Survey data to explore **developer demographics, current technology usage, and future technology preferences**.

### Objectives
- Examine respondent demographics.
- Identify commonly used programming languages, databases, platforms, and web frameworks.
- Identify technologies developers want to work with in the future.
- Compare current technology usage with future preferences.
- Produce clear visualizations and summarize key findings.

> **Note:** This GitHub version uses a 5,000-row sample of the original survey dataset so that the project can be shared on GitHub without exceeding GitHub's file-size limit. Results from this sample may differ from the full Coursera dataset.


## 1. Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
sns.set_theme(style="whitegrid")
```

## 2. Load the Dataset

```python
file_path = "../data/survey_data_sample.csv"
df = pd.read_csv(file_path)

print("Dataset shape:", df.shape)
df.head()
```

## 3. Data Overview

```python
print("Number of rows:", len(df))
print("Number of columns:", len(df.columns))

display(df.info())
display(df.isnull().sum().sort_values(ascending=False).head(15))
```

## 4. Data Cleaning and Preparation

```python
# Standardize blank strings as missing values
df = df.replace(r"^\s*$", np.nan, regex=True)

# Display missing-value percentages
missing_pct = (df.isnull().mean() * 100).sort_values(ascending=False)
display(missing_pct.to_frame("Missing %").head(15))
```

## 5. Developer Demographics

This section examines the distribution of respondents by age, country, and formal education level.


```python
# Age distribution
if "Age" in df.columns:
    age_counts = df["Age"].dropna().value_counts()
    display(age_counts.head(15))

    plt.figure(figsize=(10, 5))
    age_counts.head(10).sort_values().plot(kind="barh")
    plt.title("Top Age Groups of Survey Respondents")
    plt.xlabel("Number of Respondents")
    plt.ylabel("Age")
    plt.tight_layout()
    plt.show()
```

```python
# Country distribution
if "Country" in df.columns:
    country_counts = df["Country"].dropna().value_counts().head(10)

    plt.figure(figsize=(10, 5))
    country_counts.sort_values().plot(kind="barh")
    plt.title("Top 10 Countries by Number of Respondents")
    plt.xlabel("Number of Respondents")
    plt.ylabel("Country")
    plt.tight_layout()
    plt.show()
```

```python
# Formal education distribution
if "EdLevel" in df.columns:
    education_counts = df["EdLevel"].dropna().value_counts().head(10)

    plt.figure(figsize=(10, 6))
    education_counts.sort_values().plot(kind="barh")
    plt.title("Formal Education Levels of Respondents")
    plt.xlabel("Number of Respondents")
    plt.ylabel("Education Level")
    plt.tight_layout()
    plt.show()
```

## 6. Current Technology Usage

The following analysis identifies the most frequently reported programming languages, databases, platforms, and web frameworks currently used by respondents.

```python
def count_multiselect(column, top_n=10):
    if column not in df.columns:
        return pd.Series(dtype="int64")
    values = (
        df[column]
        .dropna()
        .astype(str)
        .str.split(";")
        .explode()
        .str.strip()
    )
    values = values[values.ne("")]
    return values.value_counts().head(top_n)

def plot_top_counts(counts, title):
    if counts.empty:
        print("No data available for this analysis.")
        return
    plt.figure(figsize=(10, 6))
    counts.sort_values().plot(kind="barh")
    plt.title(title)
    plt.xlabel("Number of Respondents")
    plt.ylabel("")
    plt.tight_layout()
    plt.show()
```

```python
current_languages = count_multiselect("LanguageHaveWorkedWith")
display(current_languages.to_frame("Respondents"))
plot_top_counts(current_languages, "Top 10 Programming Languages Currently Used")
```

```python
current_databases = count_multiselect("DatabaseHaveWorkedWith")
display(current_databases.to_frame("Respondents"))
plot_top_counts(current_databases, "Top 10 Databases Currently Used")
```

```python
current_platforms = count_multiselect("PlatformHaveWorkedWith")
display(current_platforms.to_frame("Respondents"))
plot_top_counts(current_platforms, "Top 10 Platforms Currently Used")
```

```python
current_webframes = count_multiselect("WebFrameHaveWorkedWith")
display(current_webframes.to_frame("Respondents"))
plot_top_counts(current_webframes, "Top 10 Web Frameworks Currently Used")
```

## 7. Future Technology Preferences

This section identifies the technologies respondents want to work with in the future.

```python
future_languages = count_multiselect("LanguageWantToWorkWith")
display(future_languages.to_frame("Respondents"))
plot_top_counts(future_languages, "Top 10 Programming Languages Developers Want to Work With")
```

```python
future_databases = count_multiselect("DatabaseWantToWorkWith")
display(future_databases.to_frame("Respondents"))
plot_top_counts(future_databases, "Top 10 Databases Developers Want to Work With")
```

```python
future_platforms = count_multiselect("PlatformWantToWorkWith")
display(future_platforms.to_frame("Respondents"))
plot_top_counts(future_platforms, "Top 10 Platforms Developers Want to Work With")
```

```python
future_webframes = count_multiselect("WebframeWantToWorkWith")
display(future_webframes.to_frame("Respondents"))
plot_top_counts(future_webframes, "Top 10 Web Frameworks Developers Want to Work With")
```

## 8. Current vs Future Programming Language Interest

```python
comparison = pd.DataFrame({
    "Currently Used": current_languages,
    "Want to Work With": future_languages
}).fillna(0)

comparison = comparison.sort_values("Want to Work With", ascending=False).head(10)
display(comparison)

comparison.plot(kind="bar", figsize=(12, 6))
plt.title("Current Usage vs Future Interest: Top Programming Languages")
plt.xlabel("Programming Language")
plt.ylabel("Number of Respondents")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

## 9. Key Findings

Based on this sample analysis:

- The survey contains respondents with varied demographic and educational backgrounds.
- A small group of programming languages accounts for a substantial share of reported current usage.
- Developers also show strong interest in a number of technologies they want to use in the future.
- Comparing current usage with future preferences helps identify technologies that are both established and potentially emerging.
- The dashboard visualizations provide a complementary way to explore these patterns.

**Important:** Because this GitHub notebook uses a 5,000-row sample rather than the complete dataset, these findings should be interpreted as sample-based observations rather than exact results for the full survey.
